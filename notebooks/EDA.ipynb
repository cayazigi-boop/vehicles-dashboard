```python
import pandas as pd
```

1) Lendo a tabela 'vehicles_us.csv' e passando para o data frame df:

```python
df=pd.read_csv("../vehicles_us.csv")
```

2) Vendo os tipos de dados e valores ausentes do data frame:

```python
df.info()
```

3) Verificando quantas ocorrencias de cada valor único da coluna 'is_4wd'aparecem, ignorando os valores nulos (NaN) :

```python
df['is_4wd'].value_counts()
```

3a) Verificando o número de valores ausentes:

```python
df['is_4wd'].isna().sum()
```

4) Substituindo os valores nulos (NaN) por 0.0, porque ausência de valor não significa que não é 4wd:

```python
df['is_4wd'] = df['is_4wd'].fillna(0)
```

5) Conferindo o número de valores ausentes na coluna 'is_4wd'

```python
df['is_4wd'].isna().sum()
```

6) Verificando o número de valores ausentes na coluna 'model_number':

```python
df['model_year'].isna().sum()
```

7) Tratando valores nulos da coluna 'model_year'.
Substituindo valores nulos (NaN) de cada model, pela mediana do ano:
para cada carro com 'model_year' ausente, ela busca a mediana do ano daquele modelo específico (calculada linha a linha com o transform) e preenche o valor nulo com esse valor:

```python
df['model_year'] = df['model_year'].fillna(df.groupby('model')['model_year'].transform('median'))
```

8) ver se o número de valores ausentes nessa coluna caiu para zero:

```python
df['model_year'].isna().sum()
```

9) Verificando o número de valores ausentes na coluna 'cylinders':

```python
df['cylinders'].isna().sum()
```

10) Tratando valores nulos da coluna 'cylinders'.
Substituindo valores ausentes, pela mediana:
para cada carro com 'cylinders' ausente, ela busca a mediana e preenche o valor nulo com esse valor:

```python
df['cylinders'] = df['cylinders'].fillna(df.groupby('model')['cylinders'].transform('median'))
```

11) Verificando o número de valores ausentes na coluna 'cylinders' caiu para 0:

```python
df['cylinders'].isna().sum()
```

12) Verificando o número de valores ausentes na coluna 'odometer':

```python
df['odometer'].isna().sum()
```

13) Tratando valores nulos da coluna 'odometer'.
Substituindo valores ausentes, pela mediana, agrupado por tempo ('model_year')

```python
df['odometer'] = df['odometer'].fillna(df.groupby('model_year')['odometer'].transform('median'))
```

14) Verificando o número de valores ausentes na coluna 'odometer' caiu para 0:

```python
df['odometer'].isna().sum()
```

15) Ainda retornou um valor ausente (np.int64(1)). Vamos ver a linha com os dados:

```python
df[df['odometer'].isna()]
```

16) Como ele é o único carro com esse model_year, o grupo dele no groupby só tem essa linha, e como o valor dela já é NaN, a mediana do grupo também é NaN. Por isso o fillna não conseguiu preencher. Temos algumas opções possíveis:

Preencher esse valor com a mediana geral de odometer (de todo o dataset, sem agrupar) ou remover essa linha, já que é um caso isolado/outlier. Achei melhor remover:

```python
df = df.dropna(subset=['odometer'])
```

17) Checando se removeu a linha:

```python
df['odometer'].isna().sum()
```

18) Verificando o número de valores ausentes na coluna 'paint_color':

```python
df['paint_color'].isna().sum()
```

19) Tratando valores nulos da coluna 'paint_color'.
Substituindo valores ausentes, usando a  função mode(), para preenche-los com os valores mais frequentes. O valor '0' na função mode significa para pegarmos o primeiro valor de uma coluna se há varios valores empatados.

```python
df['paint_color'] = df['paint_color'].fillna(df['paint_color'].mode()[0])
```

20) Verificando se ainda há valores ausentes:

```python
df['paint_color'].isna().sum()
```

21) Agora vou corrigir os tipos das colunas que deveriam mudar:
model_year → float64 para int64
cylinders → float64 para int64
odometer → float64 para int64
is_4wd → float64 para boolean
date_posted → object para datetime

```python
df['model_year'] = df['model_year'].astype(int)
df['cylinders'] = df['cylinders'].astype(int)
df['odometer'] = df['odometer'].astype(int)
df['is_4wd'] = df['is_4wd'].astype(bool)
df['date_posted'] = pd.to_datetime(df['date_posted'])
```

22) Confirmando que todas as colunas estão com os tipos certos:

```python
df.info()
```

23) Verificando linhas duplicadas no DataFrame:

```python
df.duplicated().sum()
```
